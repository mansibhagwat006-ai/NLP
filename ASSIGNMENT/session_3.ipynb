##TASK 1  
Question: Take three recent Instagram captions and manually create a vocabulary list of all unique words used, ignoring case and punctuation.

```python
captions = [
    "Good vibes and great memories!",
    "Enjoying good food with friends.",
    "Great day, great memories, happy vibes."
]

vocabulary = set()

for caption in captions:
    words = caption.lower().replace("!", "").replace(".", "").replace(",", "").split()
    vocabulary.update(words)

vocabulary = sorted(vocabulary)

print("Vocabulary:")
print(vocabulary)
```

##Task 2  

Question: Convert a list of short Zomato food reviews into a Bag of Words matrix using CountVectorizer.

```python
from sklearn.feature_extraction.text import CountVectorizer

reviews = [
    "Loved the pizza",
    "Service was slow",
    "Great ambience"
]

vectorizer = CountVectorizer()

bow_matrix = vectorizer.fit_transform(reviews)

print("Features:")
print(vectorizer.get_feature_names_out())

print("\nBag of Words Matrix:")
print(bow_matrix.toarray())
```

##Task 3

Question: Given an array of WhatsApp chat messages, count and display the frequency of each word. Ignore common stopwords such as the, is, and at.

```python
messages = [
    "The movie is good",
    "The movie is very good",
    "We are at the movie theatre"
]

stopwords = {"the", "is", "at"}

word_counts = {}

for message in messages:
    words = message.lower().split()

    for word in words:
        if word not in stopwords:
            word_counts[word] = word_counts.get(word, 0) + 1

print("Word Frequencies:")

for word, count in word_counts.items():
    print(word, ":", count)
```

##Task 4

Question: Use TfidfVectorizer to transform three Flipkart product reviews into TF-IDF vectors and print the TF-IDF value for the word quality in each review.

```python
from sklearn.feature_extraction.text import TfidfVectorizer

product_reviews = [
    "The product has good quality",
    "The quality of this product is excellent",
    "Poor quality and disappointing product"
]

tfidf_vectorizer = TfidfVectorizer()

tfidf_matrix = tfidf_vectorizer.fit_transform(product_reviews)

feature_names = tfidf_vectorizer.get_feature_names_out()

quality_index = list(feature_names).index("quality")

print("TF-IDF value for 'quality':")

for i, value in enumerate(
    tfidf_matrix[:, quality_index].toarray().flatten(),
    start=1
):
    print("Review", i, ":", value)
```

##Task 5

Question: List two key differences between Bag of Words and TF-IDF and give a real-world example where TF-IDF would be more useful.

**Difference 1 – Word importance**

* Bag of Words mainly counts how many times a word appears in a document.

* TF-IDF considers how important a word is in one document compared with all the other documents. Common words get lower importance.

**Difference 2 – Type of values**

* Bag of Words produces simple numerical counts such as 0, 1, 2, etc.

* TF-IDF produces weighted numerical values that represent the importance of words.

**Real-world example**

* TF-IDF can be more useful in a Flipkart product search or review system.

* For example, words like "product" and "good" may appear in many reviews. TF-IDF can give these common words less importance and give more importance to specific words such as "durable", "battery", or "waterproof".

* This can help the system understand which words are more useful for identifying a particular product or review.
