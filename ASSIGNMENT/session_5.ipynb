## Task 5 — Basic Attention Mechanism

### Prompt used with ChatGPT

> Show a basic attention calculation example in Python using three short sentences as input. Use simple numerical vectors, calculate similarity scores with a dot product, and print the attention scores.

```python
import numpy as np

def basic_attention(sentences):
    vectors = np.array([
        [1, 0, 1],
        [1, 1, 0],
        [0, 1, 1]
    ], dtype=float)

    attention_scores = np.dot(vectors, vectors.T)

    print("Input sentences:")
    for i, sentence in enumerate(sentences, start=1):
        print(f"{i}. {sentence}")

    print("\nAttention scores:")
    print(attention_scores)
    return attention_scores

sentences = [
    "I love this food",
    "The food is tasty",
    "The delivery was fast"
]

scores = basic_attention(sentences)
```

### Explanation of Attention Scores

Attention scores show how strongly one input is related to another according to the simple vector representation used in this example. A higher score means the two sentence vectors have greater similarity, so the model can give more attention to that relationship.
