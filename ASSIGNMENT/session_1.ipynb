## Task 4

Build a simple sentiment checker that asks the user to enter a movie review. If the review contains good, print Positive. If it contains bad, print Negative. Otherwise, print Neutral.

```python
review = input("Enter your movie review: ")

review = review.lower()

if "good" in review:
    print("Positive")
elif "bad" in review:
    print("Negative")
else:
    print("Neutral")
```

## Task 5

Create a spam filter function is_spam(email_text). It returns True if the email contains wiN or free, otherwise it returns False. Test it with two examples.

```python
def is_spam(email_text):
    email_text = email_text.lower()

    if "win" in email_text or "free" in email_text:
        return True
    return False


email1 = "Congratulations! You win a free gift."
email2 = "Your order has been delivered successfully."

print(is_spam(email1))
print(is_spam(email2))
```
