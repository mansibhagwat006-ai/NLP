## Task 2

Create a function to remove all punctuation marks from a Flipkart product review string using `string.punctuation`.

```python
import string

def remove_punctuation(review):
    cleaned_text = ""

    for char in review:
        if char not in string.punctuation:
            cleaned_text += char

    return cleaned_text


review = input("Enter your Flipkart product review: ")
print(remove_punctuation(review))
```

## Task 4

Download a list of English stopwords using NLTK and write a function that removes all stopwords from a WhatsApp message string.

```python
import nltk
nltk.download("stopwords")
```

```python
from nltk.corpus import stopwords

english_stopwords = set(stopwords.words("english"))

def remove_stopwords(message):
    words = message.split()
    cleaned_words = []

    for word in words:
        if word.lower() not in english_stopwords:
            cleaned_words.append(word)

    return " ".join(cleaned_words)


message = input("Enter your WhatsApp message: ")
print(remove_stopwords(message))
```

## Task 5

Use NLTK's PorterStemmer to stem each word in a list of trending Twitter hashtags and print the original and stemmed versions side by side.

```python
from nltk.stem import PorterStemmer

stemmer = PorterStemmer()

hashtags = ["#playing", "#dancing", "#eating"]

print("Original\tStemmed")

for hashtag in hashtags:
    word = hashtag[1:]
    stemmed_word = stemmer.stem(word)
    print(hashtag, "\t", "#" + stemmed_word)
```
